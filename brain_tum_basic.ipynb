```python
### Implementing EfficientNet Model 

import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.applications.efficientnet import preprocess_input
from tensorflow.keras.callbacks import ModelCheckpoint


tf.random.set_seed(42)

# normalisation
train_datagen=ImageDataGenerator(preprocessing_function=preprocess_input,
                                 rotation_range=15,zoom_range=0.15,
                                 brightness_range=[0.8,1.2])

valid_datagen=ImageDataGenerator(preprocessing_function=preprocess_input)


# making paths
train_dir=r"C:\Users\Dave Roshan\OneDrive\Desktop\brain_tum_CNN\Training"
test_dir="brain_tumour_data/Testing"


# making batches
train_data=train_datagen.flow_from_directory(directory=train_dir,
                                             batch_size=32,target_size=(224,224),
                                             class_mode="categorical",seed=42,shuffle=True)

valid_data=valid_datagen.flow_from_directory(directory=test_dir,
                                             batch_size=32,target_size=(224,224),
                                             class_mode="categorical",seed=42)


### EFF-NET MODEL

eff_model=tf.keras.applications.EfficientNetB0(
    input_shape=(224,224,3),
    include_top=False,  # not using 1000 classes
    weights='imagenet'
)


eff_model.trainable=False  # prevents weight updation to preserve pretrained knowledge (transfer learning)

x=eff_model.output # feature maps of tumor
x=tf.keras.layers.GlobalAveragePooling2D()(x)  # FM -> 1D vectors
x=tf.keras.layers.BatchNormalization()(x)
x=tf.keras.layers.Dense(128,activation="relu")(x)  # FC dense layer->128 neurons
x=tf.keras.layers.Dropout(0.4)(x) # 40 %


output_layer=tf.keras.layers.Dense(4,activation="softmax")(x)  # each neuron for each class


# everything in one pipeline
model=tf.keras.Model(inputs=eff_model.input,outputs=output_layer) 


# compling
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
              loss="categorical_crossentropy",
              metrics=['accuracy'])

# early stopping    
early_stopp=EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

# checkpoint to save best model
checkpoint = ModelCheckpoint(
    "best_effnet.keras",
    monitor="val_accuracy",
    save_best_only=True,
    verbose=1
)

# fitting the model
history=model.fit(train_data, epochs=10, validation_data=valid_data, callbacks=[early_stopp,checkpoint])

print("done")

```

```python
# saving the effnet model

model.save("effnet_91percent.keras")
print("91% model saved successfully!")
```

```python
import matplotlib.pyplot as plt

train_acc=history.history['accuracy']
val_acc=history.history['val_accuracy']

epochs=range(len(train_acc))

plt.plot(epochs, train_acc, label= "Training Accuracy")
plt.plot(epochs, val_acc, label= "Validation Accuracy")

plt.title("Train vs Val acc.")
plt.xlabel("epochs")
plt.ylabel("Accuracy")
plt.legend()

plt.show()
```

```python
### fine tuning  (phase 1 of training alr done..now fine tuning is being done)


eff_model.trainable=True  # enables training for all layers-->allows weight updation
for i in eff_model.layers[:-10]:  # freezes all layers except last 10
    i.trainable=False

# compling
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),
              loss="categorical_crossentropy",
              metrics=['accuracy'])

# early stopping    
early_stopp=EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

# fitting the model
fine_tune_history=model.fit(train_data, epochs=10, validation_data=valid_data, callbacks=[early_stopp,checkpoint])
```

```python
# saving the fine tuned model

model.save("finetuned_effnet_91.keras")
print("Fine-tuned 91% model saved!")
```

```python
import matplotlib.pyplot as plt

train_acc=fine_tune_history.history['accuracy']
val_acc=fine_tune_history.history['val_accuracy']

epochs=range(len(train_acc))

plt.plot(epochs, train_acc, label= "Training Accuracy")
plt.plot(epochs, val_acc, label= "Validation Accuracy")

plt.title("Train vs Val acc.")
plt.xlabel("epochs")
plt.ylabel("Accuracy")
plt.legend()

plt.show()
```
